### Symmetric Encryption: Two sides, One key

Symmetric Encryption is the foundation of modern secure communication.
To scramble messages, so that only people with the same key used in encryption can decrypt them.

Symmetric -> one key is used for both encryption and decryption.
Algorithms using symmetric encryption are called symmetric ciphers.
Here we specifically use Advanced Encryption Standard (AES). (the name is actually also a title, many algos competed for this title. the one that won was called Rijndael which is now aka AES)

A AES key must be made, just random bits.

(remember there are two modes of using AES., here we provide example of ECB mode, but NEVER actually use it.)

if we use 16 bytes, we'll have 128 bit key. allowing us to use AES-128

```python
import os

key = os.urandom(16)
print(key)
```

```python
# NEVER USE ECB

from cryptography.hazmat.primitives.ciphers import Cipher, algorithms, modes
from cryptography.hazmat.backends import default_backend
import os

key = os.urandom(16)
aesCipher = Cipher(algorithm=algorithms.AES(key), mode=modes.ECB(), backend=default_backend())
aesEncryptor = aesCipher.encryptor()
aesDecryptor = aesCipher.decryptor()

# update function for enc and dec always works on 16 bytes at a time. calling with fewer than 16 bytes will not produce immediate results.
aesEncryptor.update(b'alice') # no immediate result
aesEncryptor.update(b'what') # no immediate result
aesEncryptor.update(b'is') # no immediate result
aesEncryptor.update(b'going on?') # we get output after getting at least 16 bytes of data
```

Encryption on its own is not enough to make communication secure.
1. Confidentiality
2. Integrity
3. Authentication
For actual secure connection (e.g. for a website), these three conditions need to be met. (Confidentiality in this chapter.)
(These 3 are the most critical, that need to work together)

<b>Confidentiality</b>: Only people with the right key can read the data.\
<b>Integrity</b>: Only right people can alter the data.\
<b>Authentication</b>: Knowing the identity of the people we're communicating with. (to establish identity and presence)\
(CVV code of a credit card is trivial method of proving 'presence')

In Symmetric encryption, the same key is used for encryption and decryption.
