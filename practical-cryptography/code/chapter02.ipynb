Hashing: cornerstone of cryptography.
one-way function or fingerprint.

They produce repeatable, unique values for every input.
The output value provides no clue about the input that produced it.

NOTE: MD5 algorithm is deprecated and must NOT be used outside of learning purposes. (or legacy systems)



```python
import hashlib

md5hasher = hashlib.md5() # md in MD5 stands for message digest
md5hasher.hexdigest()
```

Message digest: same document, same output. it feels random, digest must not provide any clue as to what the origin of the input was.
MD5 digest creates a number that always occupies 16 bytes of memory.

```python
import hashlib

md5hasher = hashlib.md5(b'alice')
print(md5hasher.hexdigest())

print(hashlib.md5(b'bob').hexdigest())
```

Hash functions (Cryptographic hash functions and Non-cryptographic hash functions)

An example of a non-cryptographic hash function: computing whether a number is odd or even.

hash functions try to map variable (very large) number of things into a (relatively) small set of things
MD5 for ex, no matter the size of input, we end up with 16-byte number.

in discrete algebra, this means that the domain of hash function is much large than the range.

Hash functions are lossy. we lose information going from source document to hash or digest. (This is deliberate and critical).
Otherwise there would be a way to go backward from hash to document, we don't want that.

It is easy to check for 'oddness' if we have the number (input), but basically impossible to figure out which input was used to make this ouptut. (if we only had the output)

An enormous number can be computed to check if it is odd or even. 1 for odd, 0 for even.
This is called squashing. A very large input produced a small output. and consistent. (same input, same output)

The even/odd example also compresses large inputs into a fixed size space (just one bit).
Also it is lossy, we cannot tell what number was used as input just by examining the output.

Consistency, Compression, Lossiness are fundamental qualities of all hash functions (both cryptographic and non-cryptographic).
But these are for computer science.
These qualities alone are not enough to be 'secure'.

Three more qualities are required for that:
• Preimage resistance
• Second-preimage resistance
• Collision resistance

Preimage is basically the input of hash function. (in actuality, it involves more math, but that is basically it.)

Now Preimage Resistance:
Often documents are given to hash functions as input. These documents are just set of bytes, there a large integer.
we're performing mathemtaical operations on them. Therefore, preimage is just an infinite set of integer numbers.

Preimage resistance basically means, if i was given a digest without knowing the origin, i could not even find one element in the preimage for it without doing ridiculous/impossible amount of work. (The goal of preimage resistance is to make this as hard as possible and borderline impossible.)

Inverting the hash: The process of attempting to find an element in the preimage for a given output.

This is why odd/even function is not secure. If i give you even, you can say 2. A correct input was guessed by an output. But we did not know the input, yet we were able to guess one. Hence odd/even hash function is useful but not secure.

(in fact we can describe the entire preimage of odd/even hash function easily: "all even integers".)

Another example with MD5:
MD5(x) = ca8a0fb205782051bd49f02eae17c9ee
Can you guess what x is? No, impossible.
MD5 is hard to invert.

If you have a cryptographic algorithm whose only means of being broken (inverting) is brute-force, you have a good algorithm.

### Second-Preimage and Collision Resistance

Reminder: Preimage resistance means it is really hard to find a document that produces a particular digest, unless you already know one.

Second-preimage resistance: If you already have one document that produces a particular digest, it is still really hard to find a different document that produces that same digest.
(There should be no exploitable value)

### Collision Resistance

Collision resistance means that it is hard to find any two inputs that produce the same output.
A hash algorithm is resistant to collision, it is resistant to being able to purposefully create or pick any two inputs and produce the same digest.

One property that helps with this is the avalanche property. A change in input, no matter how small, creates a large and unpredictable change in the output. (ideally 50% of output bits should be altered for small input change)

If small change cause large change in output, the attacker will have difficult time to purposefully produce a collision, forcing them to revert back to brute-force.

MD5's collision resistance is not that good. (Preimage resistance is good)
(techniques have been discovered for finding collisions)

### Digestible Hash

By this point we should be able to write a python program to compute the MD5 digest of a file.
Just remember to use byte. (open the file using rb, otherwise it may be opened as a text file by default.)
(Also do not be tempted to read the data of a text file and then convert to bytes, depending on the configuartion this may lead to unexpected results)

```python
# Exercise: Write a python program that will compute the MD5 of a file.
import hashlib

md5hasher = hashlib.md5()

with open("sample.txt", mode="rb") as file:
    print("Correct file?: ", file.readline())
    
    md5hasher.update(file.read())
    digest = md5hasher.hexdigest()
    
    print(digest)
```

To reiterate: Researchers have found ways to break MD5's preimage resistance (in labs, not in a practical manner).
MD5's collision resistance is broken, but second preimage resistance is not broken.

Once a cryptographic algorithm is broken in one way, there is increased suspicion that it is broken in other ways as well.
Even though nobody has demonstrated a practical attack on MD5’s preimage or second-preimage resistance, cryptographers worry that such vulnerabilities exist.

Ubuntu distro directory also provides SHA1SUMS and SHA256SUMS.
SHA1 is also BROKEN, and must not be used in practice (its collision resistance).

Use SHA-256. (hashlib supports it)

```python
import hashlib

md5_digest = hashlib.md5(b'alice').hexdigest()
print("MD5 digest (16 bytes, 128 bits): ", md5_digest)

sha1_digest = hashlib.sha1(b'alice').hexdigest()
print("SHA1 digest (20 bytes, 160 bits): ", sha1_digest)

sha256_digest = hashlib.sha256(b'alice').hexdigest()
print("SHA256 digest (32 bytes, 256 bits): ", sha256_digest)
```

### Hash Passwords

Servers do not store users' plain password text. They always store its hash.
To log a user in, user submits their plain password through a secure channel, server hashes that plain password in memory and compares it with the hash in its database, if both hashes are the same, server will allow the user to log in.

(Also hashing your password before sending it to the server is possible, so that the server neven sees your plain password even in memory, but this approach has some problems. It is not used since you have to trust the server anyway. Using multiple authentication methods is better.)

Two factor authentication can make things worse.
Security questions also intorduce risks.
Sending codes via SMS is very dangerous, since SMS can be easily hijacked.

One very secure way, is properly deployed challenge-response hardware tokens.
(you possess them, no need to remember, cannot be guessed, or spoofed, cannot be given accidentally and cannot be forged.)

Ultimately, you need two or more authentication factor anyway for true security. (fixing passwords is not enough)

#### Hashed Password + Salt

Another way of securing passwords (after using a secure algorithm: sha256) is Salt.
Salt is a value that is mixed with the user's password before hashing.

It needs to be unique and sufficiently long. (e.g. by using os.urandom and base64.b64encode to generate a strong, random salt)

NOTE: I think Book's example for salt is wrong (duplicate code).

```python
import hashlib, os, base64

salt = base64.b64encode(os.urandom(16)) # not sure about this, book's example was a typo
password = hashlib.sha256(salt + b'alice-password').hexdigest()
print(password)
```

The username and salt should be stored along password.
These were the basics of secure password storage. But there are better algorithms with additional steps to make it even harder for attackers to invert a password.

E.g. Scrypt by Colin Percival or bcrypt or its successor Argon2.



```python
# Scrypt example

import os
from cryptography.hazmat.primitives.kdf.scrypt import Scrypt
from cryptography.hazmat.backends import default_backend

salt = os.urandom(16)
kdf = Scrypt(salt=salt, length=32, n=2**14, r=8, p=1, backend=default_backend())
key = kdf.derive(b"my great password")
print(kdf)
print(key)

# Both key and salt must be stored on disk

# After restoring from disk we can verify
kdf = Scrypt(salt=salt, length=32, n=2**14, r=8, p=1, backend=default_backend())
kdf.verify(b"my great password", key)
print("Success (exception if mismatch.)")

try:
    kdf.verify(b"wrong password", key)
    print("ok")
except Exception as error:
    print("->", error) # ?? Context was already finalized ?? What is this?
    print("Invalid credentials")
    
    
# cryptography module is a wrapper around an lower-level engine (OpenSSL for instance, default_backend param specifies this)
# this makes things work faster, because the computaitons (required for encryption and ...) are not done in python

# length param specifies how long the key will be (32 bytes in our case)
# n, r, p are tuning params, indicating how long it takes to compute and how much memory to use
# the longer the better

# recommended values are: 8 for r, 1 for p
# n could vary, for situtations when speed is needed (like websites) or more secure situtations when quick response is not needed, either way must be a power of 2
# 2 to power of 14 for interactive logins, 2 to power of 20 for more sensitive files

# these recommendations depend on context, and could change in the future
```

### Cracking Weak Passwords

(The least sophisticated method of cracking a password is brute-force)


```python
def generate(alphabet, max_len):
    if max_len <= 0: return
    for c in alphabet:
        yield c
    for c in alphabet:
        for next in generate(alphabet, max_len):
            yield c + next

# example usage: generate('ab', 2) -> 'a', 'b', 'aa', 'ab', 'ba', 'bb'

# recommended to use:
# • string.ascii_lowercase
# • string.ascii_uppercase
# • string.ascii_letters

# but encode to byte first:
import string
string.ascii_letters.encode('utf-8')
```

```python
# Exercise: (time this function later)
import string, random, hashlib

# this is an example of a brute-force function
def find_match(chars: list[str]):
    preimage_seed = string.ascii_lowercase[random.randint(0, 25)].encode('utf-8')
    digest = hashlib.md5(preimage_seed).hexdigest()
    for i in chars:
        if hashlib.md5(i.encode('utf-8')).hexdigest() == digest:
            print(f"Found match, preimage: {preimage_seed} -> digest: {digest} | character: {i}")
            return True
    return False
    

# print(find_match(['a', 'b', 'c', 'd', 'e', 'f', 'g']))

def run_find_match(n: int, chars: list[str]):
    for i in range(n):
        print(f"Iteration: {i+1}: Result -> {find_match(chars)}")
        
run_find_match(3, ['a', 'b', 'c', 'd', 'e', 'f', 'g'])

# Repeat this exercise with more and larger input (2, 3 word length inputs and ...), lowercase, uppercase, numbers and finally all printable characters: string.printable (and measure time)
# Result: increasing the length of password and size of alphabet both will increase the time it takes to invert the hash

# Now try repeating the exercise with more secure hashing algorithms: sha-1, sha-256 (scyrpt may take a long time)

# IMPORTANT: a password being just long does not make it secure. hackers have dictionaries containing common phrases
# example: chocolatecake is long but still very easily broken
```
